```python
import pandas as pd
```

```python
path = "data-model/Data Model.xlsx"
```

```python
df = pd.read_excel(path)

df["contacts"] = (
    df["contacts"]
    .astype("string")
    .str.replace("\u00A0", "", regex=False)
    .str.strip()
)
num = pd.to_numeric(df["contacts"].str.replace("+", "", regex=False), errors="coerce")
bool_cond = num.notna()

df.loc[bool_cond, "contacts"] = "+" + num.loc[bool_cond].astype("Int64").astype(str)

df["id_num"] = pd.to_numeric(df["id"], errors="coerce")

# to remove unnecessary characters from a string
df["openingHours"] = df["openingHours"].astype("string").str.replace("\u00A0", " ", regex=False).str.replace(r"\s+", " ", regex=True).str.strip()

df_locations = df[
    df["id_num"].notna() &
    df["lat"].notna() &
    df["lng"].notna()
].copy()

df_locations["id"] = df_locations["id_num"].astype(int)
df_locations = df_locations.drop(columns=["id_num", "address", "rating", "wineTags", "experienceTags"])

df_locations
```
