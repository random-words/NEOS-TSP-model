# DataFrame Preparation Function

```python
def prepare_tsp_df(df):
    df2 = df.copy()

    # Переводимо важливі колонки в числовий формат
    for col in [
        "id", "lat", "lon",
        "avgStayMinutes", "priceLevel",
        "avgTastingPricePerPerson", "avgMealPricePerPerson", "avgBottlePrice"
    ]:
        df2[col] = pd.to_numeric(df2[col], errors="coerce")

    # Беремо тільки реальні точки (де є id, lat, lon)
    df2 = df2.dropna(subset=["id", "lat", "lon"])

    # id як цілі числа
    df2["id"] = df2["id"].astype(int)

    # Для зручності – сортуємо по id
    df2 = df2.sort_values("id").reset_index(drop=True)

    target_columns = ["id", "lat", "lon", "avgStayMinutes", "priceLevel", "avgTastingPricePerPerson", "avgMealPricePerPerson", "avgBottlePrice"]
    prepared_df = df2[target_columns]

    return prepared_df
```
